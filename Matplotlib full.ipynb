```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {

    "Month" : [ 'Jan', 'Feb', 'March', 'April', 'May', 'June'],
    "Sale" :   [ 10000, 12000, 15000, 13000, 15500, 17000 ],
    "Profit" : [ 2000, 3000, 4000, 2500, 3500, 5000, ]

}

df = pd.DataFrame(data)

print(df)
```

```python
plt.figure(figsize = (7,3))
plt.plot(df['Month'], df['Sale'],  color = 'blue', marker = 'o', linestyle = '--', label = 'Sales')
plt.title('Sales Trend over Month')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.legend()
plt.show()
```

```python
plt.figure(figsize = (8,3))
width = 0.3
plt.bar(df['Month'], df['Sale'],  width= width, color = 'skyblue', label = 'Sale')
plt.bar(df['Month'], df['Profit'], width = width, color = 'green' , label = 'Profit', bottom = df['Sale'])
plt.title('Sales Trend over Month')
plt.xlabel('Month')
plt.ylabel('Amount')

plt.grid(True)
plt.tight_layout()
plt.legend()
plt.show()
    
```

```python
plt.figure(figsize = (10,5))
plt.pie(df['Profit'], labels = df['Month'] ,autopct = '%1.2f%%',startangle = 140, colors=plt.cm.Paired.colors)
plt.title('Profit by month')
plt.show()
```

```python
plt.figure(figsize = (10,5))
plt.scatter(df['Sale'], df['Profit'], color = 'green', s = 100,  edgecolors= 'black' )
plt.title('Sales vs Profit Scatter Plot')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.tight_layout()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize = (10,5))
plt.hist(df['Sale'], bins = 5, color = 'purple', edgecolor = 'black')
plt.title('Sales Distributon')
plt.xlabel('Sale')
plt.ylabel('Frequency')


plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize = (10,5))
plt.boxplot(df['Profit'], vert = False, patch_artist = True, boxprops = dict(facecolor = 'lightgreen'))
plt.title('Sales Box Plot')
plt.ylabel('Sales')
```

```python
import gradio as gr
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {

    "Month" : [ 'Jan', 'Feb', 'March', 'April', 'May', 'June'],
    "Sale" :   [ 10000, 12000, 15000, 13000, 15500, 17000 ],
    "Profit" : [ 2000, 3000, 4000, 2500, 3500, 5000, ]

}

df = pd.DataFrame(data)
```

```python
def generate_plot(plot_type):
    fig = plt.figure(figsize = (10,5))

    if plot_type == "Line Plot":
        plt.plot(df['Month'], df['Sale'],  color = 'blue', marker = 'o', linestyle = '--', label = 'Sales')
        plt.title('Sales Trend over Month')
        plt.xlabel('Month')
        plt.ylabel('Sales')
        plt.grid(True)
        plt.legend()
        plt.show(),
        
    elif plot_type == 'Stacked bar chart':
        fig.set_size_inches(10,6)
        width = 0.3
        plt.bar(df['Month'], df['Sale'],  width= width, color = 'skyblue', label = 'Sale')
        plt.bar(df['Month'], df['Profit'], width = width, color = 'green' , label = 'Profit', bottom = df['Sale'])
        plt.title('Sales Trend over Month')
        plt.xlabel('Month')
        plt.ylabel('Amount')
  
        plt.grid(True)
        plt.tight_layout()
        plt.legend(),
        

    elif plot_type == 'Pie - chart':
        fig.set_size_inches(7,7)
        plt.pie(df['Profit'], labels = df['Month'] ,autopct = '%1.2f%%',startangle = 140, colors=plt.cm.Paired.colors)
        plt.title('Profit by month'),
           

    elif plot_type == 'Scatter Plot':
        
        plt.scatter(df['Sale'], df['Profit'], color = 'green', s = 100,  edgecolors= 'black' )
        plt.title('Sales vs Profit Scatter Plot')
        plt.xlabel('Sales')
        plt.ylabel('Profit')
        plt.tight_layout()
        plt.grid(True),
           

    elif plot_type == "Histogram":

        
        plt.hist(df['Sale'], bins = 5, color = 'purple', edgecolor = 'black')
        plt.title('Sales Distributon')
        plt.xlabel('Sale')
        plt.ylabel('Frequency'),

    elif plot_type == 'Box Plot':
        plt.boxplot(df['Profit'], vert = False, patch_artist = True, boxprops = dict(facecolor = 'lightgreen'))
        plt.title('Sales Box Plot')
        plt.ylabel('Sales'),


    plt.tight_layout()
    return fig


# Gradio UI
demo = gr.Interface(
    fn=generate_plot,

    inputs=gr.Radio(
        choices=["Line Plot", 'Pie - chart', "Scatter Plot", "Histogram","Stacked bar chart", "Box Plot"],
        label="Select Visualization"
    ),

    outputs=gr.Plot(
        label="Sales Data Visualization"
    ),

    title="Sales & Profit Visual Insight",

    description="Choose the type of visualization to display the data."
)


demo.launch()


    
    
        


        
        
        
        
    
        
        
```
